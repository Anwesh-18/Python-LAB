```python
import numpy as np          # Importing the neccessary libraries
import pandas as pd
import math
from collections import Counter
```

```python
# Preparing the data set
data = {
    "x1":[0,2,0,0,-1,1],
    "x2":[3,0,1,1,0,1],
    "x3":[0,0,3,2,1,1],
    "y":['red','red','red','green','green','red'] 
}
data = pd.DataFrame(data)
data.head()
```

```python
# Separating the features and target values
X_train = data[['x1','x2','x3']].values
y_train = data['y'].values
```

```python
# Using the Euclidean function to calculate the distance
def euclidean_dist(tst,trn):
    return math.sqrt(sum((a-b)**2 for a,b in zip(tst,trn)))
```

```python
# KNN Classifying method for predicting the target value
def KNN_Classify(test,X_train,y_train,k):
    distance = []
    
    # Traversing the feature data points to calculate the distance
    for i in range(len(X_train)):
        dist = euclidean_dist(test,X_train[i])
        distance.append((dist,y_train[i]))
    
    # Sorting the distance list according to distances
    distances = sorted(distance,key=lambda x: x[0])[:k]
    # Filtering out the required labels
    req_labels = [label for (_,label) in distances]  
    
    # Counting the frequency of labels and finding out the most common one
    freq = Counter(req_labels).most_common(1)
    return freq[0][0]
```

```python
test_data = list(map(int,input("Enter the test data (X1,X2,X3): ").split()))
predicted_label = KNN_Classify(test_data,X_train,y_train,2)

print("Your test Data: ",test_data)
print("K value: ",k)
print("Predicted label: ",predicted_label)
```
