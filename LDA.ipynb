```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("iris.csv",names=["sepal_length","sepal_width","petal_length","petal_width","class"])
df.head()
```

```python
df["class"].unique()
```

```python
df["class"] = df["class"].map({'Iris-setosa':0, 'Iris-versicolor':1, 'Iris-virginica':2})
df.head()
```

```python
X = df.drop(columns=["class"]).values
y = df["class"].values
classes = df["class"].unique()
```

```python
mean_vectors = []
for c in classes:
    mean_vectors.append(np.mean(X[y==c],axis=0))
mean_vectors
```

```python
S_w = np.zeros((4,4))
for c,mean_vec in zip(classes,mean_vectors):
    class_scater = np.zeros((4,4))
    for row in X[y==c]:
        diff = (row-mean_vec).reshape(4,1)
        class_scater += diff.dot(diff.T)
    S_w += class_scater
```

```python
overall_mean = np.mean(X,axis=0).reshape(4,1)

S_B = np.zeros((4,4))
for c,mean_v in zip(classes,mean_vectors):
    n_c = X[y==c].shape[0]
    mean_v = mean_v.reshape(4,1)
    diff = mean_v - overall_mean
    S_B += n_c * diff.dot(diff.T)
```

```python
eig_vals,eig_vecs = np.linalg.eig(np.linalg.inv(S_w).dot(S_B))
```

```python
idx = np.argsort(eig_vals)[::-1]
eig_vals = eig_vals[idx]
eig_vecs = eig_vecs[:,idx]
```

```python
W = eig_vecs[:,:2]

X_lda = X.dot(W)
df_lda = pd.DataFrame(X_lda,columns=["LD1","LD2"])
df_lda["class"] = y
df_lda.head() 
```
