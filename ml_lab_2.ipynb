```python
# Simple imputer is a class which is used to handle the missing values with simple strategies
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(missing_values=np.nan,strategy="mean")  # Filling the missing values with mean
X.iloc[:,1:3] = imputer.fit_transform(X.iloc[:,1:3])
X
```

```python
X = data.iloc[:,:-1]
imputer = SimpleImputer(missing_values=np.nan,strategy="constant",fill_value=50)
X.iloc[:,1:3] = imputer.fit_transform(X.iloc[:,1:3])
X
```

```python
X = data.iloc[:,:-1]
imputer = SimpleImputer(missing_values=np.nan,strategy="median")
X.iloc[:,1:3] = imputer.fit_transform(X.iloc[:,1:3])
X
```

```python
X = data.iloc[:,:-1]
imputer = SimpleImputer(missing_values=np.nan,strategy="most_frequent")
X.iloc[:,1:3] = imputer.fit_transform(X.iloc[:,1:3])
X
```
