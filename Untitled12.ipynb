```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("YachtHydrodynamics.csv")
df.head()
```

```python
# sparating the features and target colunmns
y = df.iloc[:, 0].values
X = df.iloc[:, 1:].values

# Add intercept term (bias)
X = np.c_[np.ones(X.shape[0]), X]
```

```python
print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
```

```python
# 4. Train-Test Split (80%-20%)
np.random.seed(42)  # for reproducibility
rows = X.shape[0]   # number of rows

rp = np.random.permutation(rows) 

train_size = int(np.round(rows * 0.8))
train_idx, test_idx = rp[:train_size], rp[train_size:]

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
bt0 = np.mean(y_train)
y_train_centered = y_train - bt0
x_mean = np.mean(X_train, axis=0) 
X_train_centered = X_train - x_mean
```

```python
# 3. Computing Regression Coefficients (beta) using centered variables
XtX = np.dot(X_train_centered.T, X_train_centered)
lam = 5 * np.identity(X_train_centered.shape[1])
XtX_plus_lam = XtX + lam

XtX_plus_lam_inv = np.linalg.inv(XtX_plus_lam)
XtX_plus_lam_inv_Xt = np.dot(XtX_plus_lam_inv, X_train_centered.T)

# beta value
bt = np.dot(XtX_plus_lam_inv_Xt, y_train_centered)

print("Regression Coefficients (beta):\n", bt)
```

```python
# calculating bt0_cap
bt0_cap = bt0 - np.dot(x_mean,bt)
bt0_cap
```

```python
# predicting the final y_values
y_pred = bt0_cap + np.dot(X_test,bt)
y_pred
```
